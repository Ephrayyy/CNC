```python
px.histogram(
    df,
    x='release_year',)
```

```python
kpis = {
    'rows': len(df),
    'date_min': str(df['release_date'].min().date()),
    'date_max': str(df['release_date'].max().date()),
    'avg_rating': round(df['vote_average'].mean(), 2),
    'avg_roi': round(df['roi'].mean(), 2),
    'total_revenue_billion': round(df['revenue'].sum() / 1_000_000_000, 2),
}
kpis
```

```python
year_counts = df['release_date'].dt.year.value_counts().sort_index().reset_index()
year_counts.columns = ['release_year', 'n_titles']
px.bar(year_counts, x='release_year', y='n_titles', title='Nombre de films par an')
```

```python
language_counts = df['original_language'].value_counts().head(10).reset_index()
language_counts.columns = ['original_language', 'n_titles']
px.bar(language_counts, x='original_language', y='n_titles', title='Top 10 langues originales')
```

```python
country_counts = df['production_country'].value_counts().head(10).reset_index()
country_counts.columns = ['production_country', 'n_titles']
px.bar(country_counts, x='production_country', y='n_titles', title='Top 10 pays de production')
```

```python
genre_counter = Counter()
for value in df['genres'].dropna():
    for genre in [chunk.strip() for chunk in str(value).split(',') if chunk.strip()]:
        genre_counter[genre] += 1

genre_df = pd.DataFrame(genre_counter.items(), columns=['genre', 'count']).sort_values('count', ascending=False).head(12)
px.bar(genre_df, x='genre', y='count', title='Genres les plus fréquents')
```

```python
px.scatter(
    df,
    x='budget',
    y='revenue',
    color='original_language',
    hover_name='title',
    title='Relation budget / revenue'
)
```

```python
top_revenue = df[['title', 'release_date', 'revenue', 'budget', 'vote_average', 'popularity']].sort_values('revenue', ascending=False).head(15)
top_revenue
```

```python
top_roi = df.loc[df['budget'] > 0, ['title', 'release_date', 'budget', 'revenue', 'roi']].sort_values('roi', ascending=False).head(15)
top_roi
```
