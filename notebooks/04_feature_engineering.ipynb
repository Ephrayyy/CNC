# 04 - Feature Engineering

Ce notebook transforme la base nettoyée en table de features exploitable pour la modélisation.

```python
from pathlib import Path
import sys
import pandas as pd

ROOT_DIR = Path.cwd()
if ROOT_DIR.name == 'notebooks':
    ROOT_DIR = ROOT_DIR.parent

if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

from src.processing.feature_engineering import build_feature_table, get_model_inputs

PROCESSED_PATH = ROOT_DIR / 'data' / 'processed' / 'movies_clean.csv'
df_clean = pd.read_csv(PROCESSED_PATH, parse_dates=['release_date'])
```

```python
df_features = build_feature_table(df_clean)
print(df_features.shape)
df_features.head(3)
```

```python
new_feature_columns = [
    'release_year', 'release_month', 'release_quarter', 'is_holiday_window',
    'budget_musd', 'revenue_musd', 'profit_musd', 'vote_count_log',
    'popularity_log', 'num_genres', 'success_score', 'success_label'
]
df_features[new_feature_columns].head(10)
```

```python
X, y, feature_columns = get_model_inputs(df_features)
print('X shape =', X.shape)
print('y shape =', y.shape)
print('Feature columns =', feature_columns)
```

```python
target_distribution = y.value_counts(normalize=True).rename('share').reset_index()
target_distribution.columns = ['success_label', 'share']
target_distribution
```
